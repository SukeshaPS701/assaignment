```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import string

# Download required data
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt')

# PARAGRAPH 1 - WITH MULTIPLE PUNCTUATION MARKS
paragraph1 = """
Cloud computing has revolutionized how businesses manage their IT infrastructure!
Instead of maintaining expensive on-premise servers, companies can now use cloud
services provided by vendors such as Amazon Web Services (AWS), Microsoft Azure,
and Google Cloud Platform. Cloud computing offers several advantages: scalability,
cost-efficiency, and flexibility. The three main service models are Infrastructure
as a Service (IaaS)-which provides virtualized computing resources over the internet;
Platform as a Service (PaaS)-which offers development environments and tools; and
Software as a Service (SaaS)-which delivers applications directly to users. Organizations
ask: "How can we migrate to the cloud?" The answer requires careful planning, but the
benefits are significant-reduced capital expenditure, improved reliability, and enhanced
security. Many enterprises have already transitioned to cloud-based solutions!
"""

# PARAGRAPH 2 - WITH MULTIPLE PUNCTUATION MARKS
paragraph2 = """
Amazon Web Services (AWS) is the leading cloud platform, offering a comprehensive suite
of services. Popular AWS services include EC2 (Elastic Compute Cloud) for virtual servers,
S3 (Simple Storage Service) for scalable object storage, RDS (Relational Database Service)
for managed databases, and Lambda for serverless computing. AWS also provides services for
networking (VPC, Route 53), analytics (Redshift, Athena), machine learning (SageMaker), and
security (IAM, KMS). Companies using AWS can: reduce operational costs, improve application
performance, and scale resources on-demand-all without managing physical hardware! However,
organizations must understand AWS pricing models; they include on-demand instances, reserved
instances, and spot instances. Professionals certified in AWS (Solutions Architect, Developer
Associate) are in high demand. The cloud computing industry continues to grow rapidly-with
projections showing substantial increases in market share over the coming years!
"""

# COMBINE 2 PARAGRAPHS AS 1 INPUT
combined_text = paragraph1 + " " + paragraph2

print("\n" + "="*80)
print("TEXT PREPROCESSING ASSIGNMENT")
print("="*80)

print("\n--- ORIGINAL TEXT (2 PARAGRAPHS ON CLOUD COMPUTING & AWS WITH MULTIPLE PUNCTUATION) ---")
print(combined_text.strip())

# Step 1: Remove punctuation marks
print("\n" + "="*80)
print("STEP 1: REMOVE PUNCTUATION MARKS")
print("="*80)
print(f"\nPunctuation marks to be removed: {string.punctuation}")
print("These include: . , : ; ! ? ( ) - \" ' [ ] { } / @ # $ % ^ & * + = ~ ` |")

text_without_punct = combined_text.translate(str.maketrans('', '', string.punctuation))
print("\nText after removing punctuation:")
print(text_without_punct.strip())

# Step 2: Tokenization
print("\n" + "="*80)
print("STEP 2: TOKENIZATION")
print("="*80)

tokens = text_without_punct.lower().split()
print(f"\nTotal tokens: {len(tokens)}")
print(f"Tokens: {tokens}")

# Step 3: Remove stop words
print("\n" + "="*80)
print("STEP 3: STOP WORD REMOVAL")
print("="*80)

stop_words = set(stopwords.words('english'))
print(f"\nTotal stop words: {len(stop_words)}")

filtered_tokens = [word for word in tokens if word not in stop_words]
print(f"\nTokens after removing stop words: {len(filtered_tokens)}")
print(f"Filtered tokens: {filtered_tokens}")

# Step 4: Stemming using Porter Stemmer
print("\n" + "="*80)
print("STEP 4: STEMMING (PORTER STEMMER)")
print("="*80)

stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

print(f"\nStemmed tokens: {stemmed_tokens}")

print("\nStemming examples (Original -> Stemmed):")
print("-" * 60)
for original, stemmed in zip(filtered_tokens, stemmed_tokens):
    if original != stemmed:
        print(f"  {original:20} -> {stemmed}")

# Step 5: Lemmatization
print("\n" + "="*80)
print("STEP 5: LEMMATIZATION")
print("="*80)

lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print(f"\nLemmatized tokens: {lemmatized_tokens}")

print("\nLemmatization examples (Original -> Lemmatized):")
print("-" * 60)
for original, lemmatized in zip(filtered_tokens, lemmatized_tokens):
    if original != lemmatized:
        print(f"  {original:20} -> {lemmatized}")

# Summary
print("\n" + "="*80)
print("SUMMARY STATISTICS")
print("="*80)
print(f"\nOriginal tokens: {len(tokens)}")
print(f"After stop word removal: {len(filtered_tokens)}")
print(f"After stemming: {len(stemmed_tokens)}")
print(f"After lemmatization: {len(lemmatized_tokens)}")
print(f"Reduction rate: {((len(tokens) - len(filtered_tokens)) / len(tokens)) * 100:.2f}%")

# Final Output
print("\n" + "="*80)
print("FINAL PREPROCESSED TEXT")
print("="*80)

print("\nStemmed version:")
print(" ".join(stemmed_tokens))

print("\n\nLemmatized version:")
print(" ".join(lemmatized_tokens))
```

```python
# Import libraries
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Sample input text
corpus = [
    "NLP is amazing and fun",
    "I love learning NLP",
    "Feature engineering is important in NLP"
]

# -------------------------------
# Bag of Words (BoW)
# -------------------------------
print("----- Bag of Words -----")

bow = CountVectorizer()
X_bow = bow.fit_transform(corpus)

print("Vocabulary:", bow.get_feature_names_out())
print("BoW Matrix:\n", X_bow.toarray())


# -------------------------------
# TF-IDF
# -------------------------------
print("\n----- TF-IDF -----")

tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)

print("Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", X_tfidf.toarray())
```

```python
# =========================================================
# Assignment 1 - Text Preprocessing
# Assignment 2 - TF-IDF and Word2Vec
# =========================================================

# Install (run once if needed)
!pip install gensim nltk scikit-learn

# -------------------------------
# Import Libraries
# -------------------------------
import nltk
import re
import string
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize, sent_tokenize
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from gensim.models import Word2Vec

# Download required data
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt')
nltk.download('omw-1.4')

# =========================================================
# Assignment 1 - Text Preprocessing
# =========================================================

text = """In many traditions of thought, and in many quiet reflections of the human mind, people have asked a question that is both simple and endlessly deep: what does art point to, and why does it exist at all? When we stand before a painting, or when we listen to a piece of music, or when we read a poem that somehow touches the heart, we often feel as if, perhaps, there is something more than just color, sound, or language there. It is not only paint on a canvas; it is not only notes arranged in time; it is not merely words, commas, pauses, and punctuation marks. Rather, it seems to suggest, gently and persistently, that behind the work there is a mind, a will, a creator."""

text2 = """Now, of course, the conversation does not end there. Some might argue that art is only human expression, nothing more; it is merely the outcome of neurons firing, of culture evolving, of materials interacting. But even if that were so, the question returns again, almost stubbornly: why does the human mind itself seek meaning? Why do we, again and again, create symbols, stories, images, and sounds that reach beyond survival, beyond utility, beyond mere function? The existence of art, in this sense, raises a further question about the nature of the human person, and perhaps, quietly, about the nature of reality itself."""

# Combine text
text_combined = text + "\n" + text2

# Remove punctuation
text_clean = re.sub(f"[{re.escape(string.punctuation)}]", "", text_combined)

# Tokenization
tokens = word_tokenize(text_clean.lower())

# Stopword removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]

# Final cleaned text
final_clean_text = " ".join(filtered_tokens)

print("Tokens after cleaning:\n", filtered_tokens[:10])
print("\nClean Text:\n", final_clean_text)

# -------------------------------
# Stemming & Lemmatization
# -------------------------------
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed_output = [stemmer.stem(word) for word in filtered_tokens]
lemmatized_output = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print(f"\n{'Original':<15} | {'Stemmed':<15} | {'Lemmatized'}")
print("-" * 45)

for i in range(15):
    print(f"{filtered_tokens[i]:<15} | {stemmed_output[i]:<15} | {lemmatized_output[i]}")

# =========================================================
# Assignment 2 - Bag of Words & TF-IDF
# =========================================================

# Better input (2 documents instead of 1)
input_data = [text, text2]

# Bag of Words
print("\n----- Bag of Words -----")
bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(input_data)

bow_df = pd.DataFrame(bow_matrix.toarray(), columns=bow_vectorizer.get_feature_names_out())
print(bow_df.iloc[:, :20])

# TF-IDF
print("\n----- TF-IDF -----")
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(input_data)

tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
print(tfidf_df.iloc[:, :20])

# =========================================================
# Assignment 2 - Word2Vec (>200 words input)
# =========================================================

long_text = """Charles Babbage was an English polymath, mathematician, and inventor.
He is considered the father of the computer because he designed the first mechanical computer.
His analytical engine contained concepts used in modern computing.
Babbage also worked on difference engines and computational mathematics.
He contributed to engineering, science, and economics.
His ideas influenced modern computer science.
The concept of programmable machines came from his work.
Today, computers are used in artificial intelligence, machine learning, and data science.
Machine learning and artificial intelligence are important in modern technology.
Computers help solve complex problems and improve human life.
"""

# Split into sentences (IMPORTANT FIX)
sentences = sent_tokenize(long_text.lower())

# Convert into words
data = [sentence.split() for sentence in sentences]

# Train Word2Vec
model = Word2Vec(data, vector_size=100, window=5, min_count=1, workers=4)

# Similarity checks
print("\n----- Word2Vec Similarity -----")

print("Similarity between 'computer' and 'engine':",
      model.wv.similarity('computer', 'engine'))

print("Similarity between 'machine' and 'learning':",
      model.wv.similarity('machine', 'learning'))

print("Similarity between 'babbage' and 'computer':",
      model.wv.similarity('babbage', 'computer'))

print("\nMost similar words to 'computer':")
print(model.wv.most_similar("computer", topn=3))
```
