```python
import numpy as np
import nltk

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.svm import SVC
from sklearn.cluster import KMeans

nltk.download('stopwords')
```

```python
text = """Artificial intelligence and machine learning are transforming the world.
Machine learning helps systems learn from data and improve over time."""

# Lowercase
text = text.lower()

# Tokenization
tokens = text.split()

# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered = [word for word in tokens if word not in stop_words]

print("Filtered Words:", filtered)
```

```python
corpus = [text]

cv = CountVectorizer()
X = cv.fit_transform(corpus)

print("BoW Features:", cv.get_feature_names_out())
print("BoW Matrix:\n", X.toarray())
```

```python
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)

print("TF-IDF Features:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", X_tfidf.toarray())
```

```python
sentences = [
    "nlp is interesting",
    "machine learning is interesting",
    "math is boring",
    "deep learning is powerful",
    "i love studying",
    "i hate studying"
]

tokenized = [s.lower().split() for s in sentences]

model = Word2Vec(tokenized, vector_size=50, window=3, min_count=1)

print("Vector for 'machine':")
print(model.wv['machine'])
```

```python
def sent_vec(sentence):
    words = sentence.lower().split()
    vectors = []

    for word in words:
        if word in model.wv:
            vectors.append(model.wv[word])

    # Handle empty case
    if len(vectors) == 0:
        return np.zeros(model.vector_size)

    return np.mean(vectors, axis=0)
```

```python
s1 = sent_vec("nlp is interesting")
s2 = sent_vec("machine learning is interesting")
s3 = sent_vec("math is boring")

print("Similarity s1 vs s2:", cosine_similarity([s1], [s2])[0][0])
print("Similarity s1 vs s3:", cosine_similarity([s1], [s3])[0][0])
print("Similarity s2 vs s3:", cosine_similarity([s2], [s3])[0][0])
```

```python
X = [
    sent_vec("nlp is interesting"),
    sent_vec("machine learning is powerful"),
    sent_vec("i hate studying"),
    sent_vec("deep learning is boring"),
    sent_vec("math is boring"),
    sent_vec("machine learning is advanced"),
    sent_vec("i love studying"),
    sent_vec("artificial intelligence is boring")
]

y = [1,1,0,0,0,1,1,0]

svm = SVC()
svm.fit(X, y)

test = sent_vec("deep learning is interesting")
print("Prediction:", svm.predict([test]))
```
